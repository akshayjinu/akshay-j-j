```python
# finding the quartile of age
q1=df['age'].quantile(0.25)
q3=df['age'].quantile(0.75)
IQR=q3-q1
lower_limit=q1-1.5*IQR
upper_limit=q3+1.5*IQR
```

```python
#clipping
df['age']=np.where(df['age']>upper_limit,upper_limit,np.where(df['age']<lower_limit,lower_limit,df['age']))
```

## Modelling

```python
# now split our data into target and feature
x=df.drop(['selection'],axis=1)
y=df['selection']
```

```python
# we need to split into train and test
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

### Logistic regression

```python
# to know whether the player selected or not
from sklearn.linear_model import LogisticRegression
```

```python
log_reg=LogisticRegression()
log_reg_model=log_reg.fit(X_train,y_train)
```

```python
log_pred=log_reg_model.predict(X_test)
```

```python
log_pred
```

```python
# Now import accuracy metrices
from sklearn.metrics import accuracy_score,confusion_matrix,accuracy_score,f1_score,precision_score,recall_score
```

```python
print('confusion matrix is :\n',confusion_matrix(y_test,log_pred))
```

```python
# checking score of confusion metrices(Before scaling)
print('accuracy is:',accuracy_score(y_test,log_pred))
print('precision is:',precision_score(y_test,log_pred))
print('recall is:',recall_score(y_test,log_pred))
print('f1 score is:',f1_score(y_test,log_pred))
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
```

```python
Scaled_xtrain=scaler.fit_transform(X_train)
Scaled_xtest=scaler.transform(X_test)
```

```python
log_reg.fit(Scaled_xtrain,y_train)
log_pred=log_reg.predict(Scaled_xtest)
```

```python
print('confusion matrix is :\n',confusion_matrix(y_test,log_pred))
```

```python
# checking score of confusion metrices(After scaling)
print('accuracy is:',accuracy_score(y_test,log_pred))
print('precision is:',precision_score(y_test,log_pred))
print('recall is:',recall_score(y_test,log_pred))
print('f1 score is:',f1_score(y_test,log_pred))
```

after scaling accuracy,precision,recall,f1 score improved alot
